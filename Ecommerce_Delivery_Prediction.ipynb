```python
# ==============================
# STEP 1: IMPORT LIBRARIES
# ==============================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

print("All libraries imported successfully!")
```

```python
# ==============================
# STEP 2: LOAD DATASET
# ==============================

df = pd.read_csv("E_Commerce.csv")

print("Dataset loaded successfully!")

print("\nDataset Shape:")
print(df.shape)

print("\nFirst 5 Rows:")
display(df.head())
```

```python
# ==============================
# STEP 3: UNDERSTAND DATASET
# ==============================

print("Dataset Information:")
print("-" * 50)
df.info()

print("\nColumn Names:")
print(df.columns.tolist())

print("\nStatistical Summary:")
display(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# ==============================
# STEP 4: TARGET VARIABLE ANALYSIS
# ==============================

target = "Reached.on.Time_Y.N"

print("Target Counts:")
print(df[target].value_counts())

print("\nTarget Percentages:")
print((df[target].value_counts(normalize=True) * 100).round(2))

plt.figure(figsize=(6, 4))

sns.countplot(
    data=df,
    x=target
)

plt.title("Distribution of Delivery Target")
plt.xlabel("Delivery Class")
plt.ylabel("Number of Orders")

plt.show()
```

```python
# ==============================
# STEP 5: CATEGORICAL EDA
# ==============================

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

sns.countplot(
    data=df,
    x="Mode_of_Shipment",
    hue="Reached.on.Time_Y.N",
    ax=axes[0, 0]
)
axes[0, 0].set_title("Mode of Shipment vs Delivery Target")

sns.countplot(
    data=df,
    x="Product_importance",
    hue="Reached.on.Time_Y.N",
    ax=axes[0, 1]
)
axes[0, 1].set_title("Product Importance vs Delivery Target")

sns.countplot(
    data=df,
    x="Warehouse_block",
    hue="Reached.on.Time_Y.N",
    ax=axes[1, 0]
)
axes[1, 0].set_title("Warehouse Block vs Delivery Target")

sns.countplot(
    data=df,
    x="Gender",
    hue="Reached.on.Time_Y.N",
    ax=axes[1, 1]
)
axes[1, 1].set_title("Gender vs Delivery Target")

plt.tight_layout()
plt.show()
```

```python
# ==============================
# STEP 6: NUMERICAL EDA
# ==============================

features = [
    "Cost_of_the_Product",
    "Discount_offered",
    "Weight_in_gms",
    "Customer_care_calls",
    "Prior_purchases",
    "Customer_rating"
]

fig, axes = plt.subplots(2, 3, figsize=(16, 9))

for feature, ax in zip(features, axes.flatten()):

    sns.boxplot(
        data=df,
        x="Reached.on.Time_Y.N",
        y=feature,
        ax=ax
    )

    ax.set_title(f"{feature} vs Delivery Target")

plt.tight_layout()
plt.show()
```

```python
# ==============================
# STEP 7: CORRELATION ANALYSIS
# ==============================

numeric_df = df.select_dtypes(include=np.number)

correlation_matrix = numeric_df.corr()

plt.figure(figsize=(11, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()

print("\nCorrelation with Delivery Target:")
print(
    correlation_matrix["Reached.on.Time_Y.N"]
    .sort_values(ascending=False)
)
```

```python
# ==============================
# STEP 8: DATA PREPROCESSING
# ==============================

# Separate features and target
X = df.drop("Reached.on.Time_Y.N", axis=1)
y = df["Reached.on.Time_Y.N"]

# Remove ID because it is only an identifier
X = X.drop("ID", axis=1)

# Convert categorical variables into numeric dummy variables
X = pd.get_dummies(
    X,
    columns=[
        "Warehouse_block",
        "Mode_of_Shipment",
        "Product_importance",
        "Gender"
    ],
    drop_first=True,
    dtype=int
)

print("Data preprocessing completed!")

print("\nShape of X:")
print(X.shape)

print("\nTarget Shape:")
print(y.shape)

print("\nProcessed Columns:")
print(X.columns.tolist())

print("\nFirst 5 Processed Rows:")
display(X.head())

print("\nData Types:")
print(X.dtypes)
```

```python
# ==============================
# STEP 9: TRAIN-TEST SPLIT
# ==============================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Train-test split completed!")

print("\nX_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts(normalize=True).round(3))

print("\nTesting Target Distribution:")
print(y_test.value_counts(normalize=True).round(3))
```

```python
# ==============================
# STEP 10: FEATURE SCALING
# ==============================

scaler = StandardScaler()

# Fit scaler only on training data
X_train_scaled = scaler.fit_transform(X_train)

# Use the same scaler on test data
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed!")

print("\nOriginal Training Shape:")
print(X_train.shape)

print("\nScaled Training Shape:")
print(X_train_scaled.shape)

print("\nScaled Testing Shape:")
print(X_test_scaled.shape)
```

```python
# ==============================
# STEP 11: LOGISTIC REGRESSION
# ==============================

lr_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

lr_model.fit(X_train_scaled, y_train)

y_pred_lr = lr_model.predict(X_test_scaled)

print("Logistic Regression trained successfully!")

print(
    "Accuracy:",
    round(accuracy_score(y_test, y_pred_lr), 4)
)
```

```python
# ==============================
# STEP 12: DECISION TREE
# ==============================

dt_model = DecisionTreeClassifier(
    random_state=42
)

dt_model.fit(X_train, y_train)

y_pred_dt = dt_model.predict(X_test)

print("Decision Tree trained successfully!")

print(
    "Accuracy:",
    round(accuracy_score(y_test, y_pred_dt), 4)
)
```

```python
# ==============================
# STEP 13: RANDOM FOREST
# ==============================

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

y_pred_rf = rf_model.predict(X_test)

print("Random Forest trained successfully!")

print(
    "Accuracy:",
    round(accuracy_score(y_test, y_pred_rf), 4)
)
```

```python
# ==============================
# STEP 14: K-NEAREST NEIGHBORS
# ==============================

knn_model = KNeighborsClassifier(
    n_neighbors=5
)

knn_model.fit(X_train_scaled, y_train)

y_pred_knn = knn_model.predict(X_test_scaled)

print("KNN trained successfully!")

print(
    "Accuracy:",
    round(accuracy_score(y_test, y_pred_knn), 4)
)
```

```python
# ==============================
# STEP 15: MODEL EVALUATION
# ==============================

predictions = {
    "Logistic Regression": y_pred_lr,
    "Decision Tree": y_pred_dt,
    "Random Forest": y_pred_rf,
    "KNN": y_pred_knn
}

for name, prediction in predictions.items():

    print("=" * 70)
    print(name)
    print("=" * 70)

    print(
        "Accuracy:",
        round(accuracy_score(y_test, prediction), 4)
    )

    print("\nConfusion Matrix:")
    print(confusion_matrix(y_test, prediction))

    print("\nClassification Report:")
    print(classification_report(y_test, prediction))

    print("\n")
```

```python
# ==============================
# STEP 16: MODEL COMPARISON
# ==============================

results = []

for name, prediction in predictions.items():

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, prediction),
        "Precision": precision_score(y_test, prediction),
        "Recall": recall_score(y_test, prediction),
        "F1 Score": f1_score(y_test, prediction)
    })

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Accuracy",
    ascending=False
).reset_index(drop=True)

print("Model Performance Comparison:")
display(results_df.round(4))
```

```python
# ==============================
# STEP 17: MODEL COMPARISON GRAPH
# ==============================

plt.figure(figsize=(9, 5))

sns.barplot(
    data=results_df,
    x="Model",
    y="Accuracy"
)

plt.title("Machine Learning Model Accuracy Comparison")
plt.xlabel("Model")
plt.ylabel("Accuracy")
plt.ylim(0, 1)

plt.xticks(rotation=15)

plt.tight_layout()
plt.show()

print("Best Model Based on Accuracy:")
print(results_df.iloc[0]["Model"])

print(
    "Best Accuracy:",
    round(results_df.iloc[0]["Accuracy"], 4)
)
```

```python
# ==============================
# STEP 18: CONFUSION MATRICES
# ==============================

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

for ax, (name, prediction) in zip(
    axes.flatten(),
    predictions.items()
):

    cm = confusion_matrix(y_test, prediction)

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        ax=ax
    )

    ax.set_title(name)
    ax.set_xlabel("Predicted")
    ax.set_ylabel("Actual")

plt.suptitle(
    "Confusion Matrices",
    fontsize=16
)

plt.tight_layout()
plt.show()
```

```python
# ==============================
# STEP 19: FEATURE IMPORTANCE
# ==============================

feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": rf_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
).reset_index(drop=True)

print("Feature Importance:")
display(feature_importance)

plt.figure(figsize=(10, 7))

sns.barplot(
    data=feature_importance,
    x="Importance",
    y="Feature"
)

plt.title("Random Forest Feature Importance")
plt.xlabel("Importance")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()
```

```python
# ==============================
# STEP 20: FINAL PROJECT SUMMARY
# ==============================

best_model = results_df.iloc[0]

print("=" * 60)
print("E-COMMERCE PRODUCT DELIVERY PREDICTION")
print("FINAL MODEL SUMMARY")
print("=" * 60)

print(f"\nBest Model: {best_model['Model']}")

print(
    f"Accuracy: {best_model['Accuracy']:.4f}"
)

print(
    f"Precision: {best_model['Precision']:.4f}"
)

print(
    f"Recall: {best_model['Recall']:.4f}"
)

print(
    f"F1 Score: {best_model['F1 Score']:.4f}"
)

print("\nTop 5 Important Features:")

for index, row in feature_importance.head(5).iterrows():

    print(
        f"{index + 1}. {row['Feature']} "
        f"({row['Importance']:.4f})"
    )

print("\nProject completed successfully!")
```

```python
# CELL 21: SAVE TRAINED MODEL

import joblib
import sklearn

model_package = {
    "model": rf_model,
    "features": X.columns.tolist(),
    "sklearn_version": sklearn.__version__
}

joblib.dump(model_package, "delivery_prediction_model.pkl")

print("Model saved successfully!")
print("File: delivery_prediction_model.pkl")
print("Scikit-learn version:", sklearn.__version__)
```
